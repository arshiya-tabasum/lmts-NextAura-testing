```python
num1 = int(input("Enter your number1 = "))
num2 = int(input("Enter your number2 = "))


op  = input("Enter your operator Like{ +, -, *, /, %, **} = ")
def calci(num1, num2, op):
    
    if op=='+':
        return num1+num2
        
       

    elif op == '-':
        return num1-num2
        

    elif op == '*':
        return num1 * num2


    elif op == '/':
        if num2 != 0:
            return num1 / num2
        else:
            return -1    
        

    elif op == '%':
        return num1 % num2
        

    elif op == "**":
        return num1 ** num2
        

num3 = calci(num1, num2, op)
print(num3)
while 1:
    option = input("Do you need to Perform any operator :(Yes/No) = ")
    if option =="Yes":
        num4 = int(input("Enter your number = "))
        ch = input("Enter your operator = ")
        print(calci(num3,num4,ch))

    else:
        print("Thank you")
        break
```
